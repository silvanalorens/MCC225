### Evaluación y ablaciones

Las métricas de este proyecto son **operativas**. Sirven para comparar condiciones dentro del mismo pipeline, no sustituyen una evaluación externa completa.

```python
from pathlib import Path
import sys
import json
import pandas as pd

PROJECT_ROOT = Path("..").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.io_utils import jsonl_load, jsonl_dump

from src.retrieval import fit_text_retriever
from src.generation import grounded_generate_template
from src.metrics import evaluate_outputs

records = jsonl_load(PROJECT_ROOT / "data_processed/records_master.jsonl")
retriever = fit_text_retriever(records)
outputs = [grounded_generate_template(r, retriever, top_k=2) for r in records]
metrics = evaluate_outputs(records, outputs)
pd.DataFrame(metrics)
```

#### Ablaciones mínimas

1. solo texto catalográfico
2. texto + estructura
3. texto + estructura + recuperación
4. texto + estructura + recuperación + plausibilidad

```python
# Un resumen ilustrativo de ablación ligera.
summary = pd.DataFrame(metrics).mean(numeric_only=True).to_frame("mean").reset_index()
summary
```
