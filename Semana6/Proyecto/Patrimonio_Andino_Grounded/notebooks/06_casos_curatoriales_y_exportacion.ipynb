### Casos curatoriales y exportación

Este cuaderno cierra el proyecto con dos lecturas cercanas: un khipu con estructura cuantitativa rica y un fragmento Paracas con alta densidad técnica e iconográfica.

```python
from pathlib import Path
import sys
import json
import pandas as pd

PROJECT_ROOT = Path("..").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.io_utils import jsonl_load, jsonl_dump

from src.retrieval import fit_text_retriever
from src.generation import grounded_generate_template

records = jsonl_load(PROJECT_ROOT / "data_processed/records_master.jsonl")
retriever = fit_text_retriever(records)

selected_ids = {"kh_0082", "par_1935_32_0212"}
selected = [r for r in records if r["canonical_id"] in selected_ids]
case_outputs = [grounded_generate_template(r, retriever, top_k=2) for r in selected]

pd.DataFrame([
    {
        "canonical_id": r["canonical_id"],
        "title": r["title"],
        "caption_factual": o["caption_factual"],
        "nota_tecnico_curatorial": o["nota_tecnico_curatorial"],
    }
    for r, o in zip(selected, case_outputs)
])
```

```python
# Exporta el paquete de casos para su discusión o evaluación.
rows = []
for r, o in zip(selected, case_outputs):
    rows.append({"canonical_id": r["canonical_id"], "title": r["title"], **o})

pd.DataFrame(rows).to_csv(PROJECT_ROOT / "outputs/evaluacion/casos_curatoriales_extensos.csv", index=False)
print("Saved:", PROJECT_ROOT / "outputs/evaluacion/casos_curatoriales_extensos.csv")
```
