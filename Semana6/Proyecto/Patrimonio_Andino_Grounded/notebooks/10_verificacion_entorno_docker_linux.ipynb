### Verificación de entorno para Docker/Linux

Este cuaderno comprueba que la carpeta del proyecto está en una ruta utilizable dentro del contenedor y que los archivos mínimos del proyecto existen.


```python
# Detectar la raíz del proyecto y mostrar el directorio de trabajo.
from pathlib import Path
import sys
import json

NOTEBOOK_DIR = Path.cwd().resolve()
PROJECT_ROOT = NOTEBOOK_DIR.parent if NOTEBOOK_DIR.name == "notebooks" else NOTEBOOK_DIR
print("Directorio actual:", NOTEBOOK_DIR)
print("Raíz detectada del proyecto:", PROJECT_ROOT)
```

```python
# Añade la raíz al path para poder importar utilidades del proyecto.
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.runtime_env import collect_runtime_report

report = collect_runtime_report(PROJECT_ROOT)
print(json.dumps(report, ensure_ascii=False, indent=2))
```

```python
# Verifica la existencia de carpetas y archivos críticos.
required = [
    PROJECT_ROOT / "data_processed" / "records_master.jsonl",
    PROJECT_ROOT / "notebooks" / "07_demo_integrada_y_modelo_real_opcional.ipynb",
    PROJECT_ROOT / "src" / "generation.py",
    PROJECT_ROOT / "configs" / "docker_linux.yaml",
]
status = {str(path.relative_to(PROJECT_ROOT)): path.exists() for path in required}
print(json.dumps(status, ensure_ascii=False, indent=2))
```
